## Step 2 — Sentiment analysis (real AI!)
Remember the Day 1 sentiment sorter you wrote by hand? Here's the same job done by
a model trained on millions of examples.

```python
classifier = pipeline("sentiment-analysis")

result = classifier("I absolutely love this course!")
print(result)
```

The result is a **list** containing a **dictionary** — the same structures
from Day 1! `label` is the verdict, `score` is the confidence (0–1).

```python
result = classifier("This is the worst thing ever.")
print("Label:", result[0]["label"])
print("Score:", result[0]["score"])
```

### -> YOUR TURN (1) — build a `feeling()` function
Wrap the classifier so it returns **just the label** for any text.

```python
def feeling(text):
    result = classifier(text)
    return result[0]["____"]      # 👉 return just the label

print(feeling("What a fantastic day!"))   # POSITIVE
print(feeling("I am so disappointed."))   # NEGATIVE
```

## Step 3 — Zero-shot classification (your OWN labels)
Classify text into categories **you** choose — no training needed.

```python
topic = pipeline("zero-shot-classification",
                 model="typeform/distilbert-base-uncased-mnli")

text = "The delivery was late and the box was damaged."
labels = ["shipping", "product quality", "price", "customer service"]

result = topic(text, labels)
print("Best match:", result["labels"][0])
```

## Step 4 — Text generation (the model writes)
The text you pass in is the **prompt**; the model continues it.

```python
generator = pipeline("text-generation", model="distilgpt2")

out = generator("Artificial intelligence will",
                max_length=30, num_return_sequences=1)
print(out[0]["generated_text"])
```

> `distilgpt2` is small and a bit quirky — perfect for learning. Bigger models write more fluently.

**B. Summarize a paragraph**

```python
summarizer = pipeline("summarization")
text = ("Python is a popular programming language. It is used for web "
        "development, data science, automation, and artificial intelligence. "
        "Its simple syntax makes it a great first language for beginners.")
print(summarizer(text, max_length=30, min_length=10)[0]["summary_text"])
```
