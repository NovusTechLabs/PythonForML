## Step 1 — NumPy warm-up
One operation applies to the whole array, no loop needed.

```python
import numpy as np

heights = np.array([150, 160, 170, 180])
print(heights + 5)        # adds 5 to every item
print(heights.mean())     # 165.0
print(heights.max())      # 180
```

## Step 2 — Load the dataset
`pd.read_csv` reads a CSV straight from a web link. Run this.

```python
import pandas as pd

url = "https://raw.githubusercontent.com/mwaskom/seaborn-data/master/penguins.csv"
df = pd.read_csv(url)

print("Loaded! Shape:", df.shape)   # (rows, columns)
```
