## Step 3 — Ask the user a question

## Step 4 — A little maths
Text typed by a user is a *string*. To do maths, convert it to a number with `int()`.

```python
age_text = input("How old are you? ")
age = int(age_text)            # turn the text into a number

print(f"In 5 years you will be {age + 5} years old.")
```
