## Step 1 — Set up the AI tools
Two pipelines: one for *how* they feel, one for *what* they mention.

```python
!pip install -q transformers
from transformers import pipeline

sentiment = pipeline("sentiment-analysis")
topic     = pipeline("zero-shot-classification",
                     model="typeform/distilbert-base-uncased-mnli")

CATEGORIES = ["shipping", "product quality", "price", "customer service"]
print("Tools ready!")
```

## Step 2 — The core function
`analyze_review()` takes text and **returns a tidy dictionary**. Complete it.

```python
def analyze_review(text):
    mood = sentiment(text)[0]["label"]              # POSITIVE / NEGATIVE
    category = topic(text, CATEGORIES)["labels"][0] # best-matching topic
    return {
        "review": text,
        "sentiment": mood,
        "topic": ____,                              # 👉 put category here
    }

print(analyze_review("The price is great but delivery was slow."))
```

## Step 3 — A readable report

```python
def print_report(result):
    print("Review :", result["review"])
    print("Mood   :", result["sentiment"])
    print("Topic  :", result["topic"])
    print("-" * 40)

print_report(analyze_review("Support was rude and unhelpful."))
```

## Step 4 — Run it over many reviews (loops!)

```python
reviews = [
    "Fast shipping and a fair price!",
    "The product broke after one day.",
    "Customer service ignored my emails.",
]

for r in reviews:
    print_report(analyze_review(r))
```
