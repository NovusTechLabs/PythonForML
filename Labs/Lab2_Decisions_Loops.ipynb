## Step 6 — Build the Sentiment Sorter 
We have keyword lists and a set of messages. For each message we check whether it
contains positive or negative words, then label it. **Run this starter:**

```python
positive_words = ["great", "good", "love", "excellent", "happy"]
negative_words = ["terrible", "bad", "hate", "awful", "sad"]

messages = [
    "great product and good value",
    "terrible service, awful experience",
    "the weather is cloudy today",
    "I love this, excellent work",
]

for msg in messages:
    label = "neutral"
    for word in positive_words:
        if word in msg:
            label = "positive"
    for word in negative_words:
        if word in msg:
            label = "negative"
    print(f"'{msg}'  ->  {label}")
```

## Stretch goals (optional)

**A. Classify a message you type yourself**

```python
my_msg = input("Type a message: ").lower()

label = "neutral"
for word in positive_words:
    if word in my_msg:
        label = "positive"
for word in negative_words:
    if word in my_msg:
        label = "negative"

print(f"That message looks {label}.")
```

**B. A 'strongly positive' tier** — count how many positive words appear and upgrade the label.

```python
msg = "great, excellent, and i love it"
hits = 0
for word in positive_words:
    if word in msg:
        hits = hits + 1

if hits >= 2:
    print("strongly positive")
elif hits == 1:
    print("positive")
else:
    print("not positive")
```
