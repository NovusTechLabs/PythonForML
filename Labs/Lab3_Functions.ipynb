`return` hands a value back so you can store and use it.

```python
def add(a, b):
    return a + b

result = add(3, 4)
print(result)        # 7
print(result * 10)   # 70
```

These are small reusable tools. Run them, then add your own.

```python
def shout(text):
    return text.upper() + "!"

def word_count(text):
    return len(text.split())   # split() makes a list of words

print(shout("hello"))            # HELLO!
print(word_count("AI is fun"))   # 3
```

## Part B — Number Guessing Game 
### Step 4: randomness from a library
`import random` gives us ready-made tools. Run this a few times.

```python
import random

print(random.randint(1, 6))                  # dice roll 1-6
print(random.choice(["yes", "no", "maybe"])) # random pick
```

### Step 5: the game
The computer picks a secret number; you guess until correct. **Run it and play!**

```python
import random

secret = random.randint(1, 20)
print("I'm thinking of a number between 1 and 20.")

while True:
    guess = int(input("Your guess: "))
    if guess == secret:
        print("🎉 Correct! Well done.")
        break               # leave the loop
    elif guess < secret:
        print("Too low, try again.")
    else:
        print("Too high, try again.")
```

> `while True:` repeats forever until `break` stops it. `break` exits the loop.

Add a `difficulty` function that sets the range: easy = 1-10, hard = 1-100.
