## Step 1 — Load the data

```python
import pandas as pd
import matplotlib.pyplot as plt

url = "https://raw.githubusercontent.com/mwaskom/seaborn-data/master/penguins.csv"
df = pd.read_csv(url)
print("Shape:", df.shape)
```

## Step 2 — Find missing values
Missing entries show up as `NaN`. Count them per column.

```python
df.isnull().sum()
```

## Step 3 — Clean: drop missing rows
We'll keep only complete rows.

```python
clean = df.dropna()
print("Before:", df.shape, " After:", clean.shape)
```

## Chart 1 — Histogram (how is body mass spread out?)

```python
plt.hist(clean["body_mass_g"], bins=20, color="#0E9594", edgecolor="white")
plt.title("Distribution of body mass")
plt.xlabel("body_mass_g")
plt.ylabel("count")
plt.show()
```

## Chart 2 — Scatter (do flipper length and body mass relate?)
We'll colour each species differently with Seaborn.

```python
import seaborn as sns

sns.scatterplot(data=clean, x="flipper_length_mm",
                y="body_mass_g", hue="species")
plt.title("Flipper length vs body mass")
plt.show()
```

> Notice the three species form **separate clusters** — a strong hint that a
model could tell them apart. That's exactly what we do next session!
